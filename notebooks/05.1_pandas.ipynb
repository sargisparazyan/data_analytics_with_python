###  P1, Analytical questions after the first merge(products_departments)

```python

import pandas as pd
df_products = pd.read_csv('../data/raw/products.csv')
df_products.head()
```

```python
df_departments_t = pd.read_csv('../data/processed/departments_t.csv')
df_departments_t.head()
```

```python
# Creating our first merge df for answering questions
df_products_departments = pd.merge(
    df_products,
    df_departments_t,
    on="department_id",
    how="left"
)

df_products_departments.head()
```

```python
# Question 1: How many products belong to each department?
df_products_departments["department"].value_counts()
```

```python
# Question 2: Which departments contain the most products?
df_products_departments["department"].value_counts().head(10)
```

```python
# Question 3: Are products concentrated in a small number of departments?
df_products_departments["department"].value_counts().head(5)
```

```python
df_aisles = pd.read_csv('../data/raw/aisles.csv')
df_aisles.head()
```

```python
# Creating our secong merge df for answering questions
df_products_enriched = pd.merge(
    df_products_departments,
    df_aisles,
    on="aisle_id",
    how="left"
)

df_products_enriched.head()
```

```python
# Question 1: How many products belong to each department?
df_products_enriched["department"].value_counts()
```

```python
# Question 2 How many products belong to each aisle?
df_products_enriched["aisle"].value_counts()
```

```python
# Question 3 Which aisles are most populated?
df_products_enriched["aisle"].value_counts().head(10)
```

```python
# Question 4: Which departments span the widest variety of products?
df_products_enriched.groupby("department")["product_id"].nunique().sort_values(ascending=False)
```

```python
# Question 5: How are departments and aisles related?
df_products_enriched.groupby("department")["aisle"].nunique().sort_values(ascending=False)
```

### P3: Questions unlocked after this merge(train_transactions)


```python
df_orders_train = pd.read_csv('../data/processed/orders_train.csv')
df_orders_train.head()
```

```python
df_order_products_train = pd.read_csv('../data/raw/order_products_train.csv')
df_order_products_train.head()
```

```python
# Merging our third dataframe(train_transactions)
df_train_transactions = pd.merge(
    df_orders_train,
    df_order_products_train,
    on = "order_id",
    how = "left"
)
df_train_transactions.head()
```

```python
# Question 1: How many products are included in each order?
products_per_order = df_train_transactions.groupby("order_id")["product_id"].count().reset_index()
products_per_order.columns = ["order_id", "product_count"]
products_per_order.value_counts()
```

```python
# Question 2: What is the typical basket size?
df_train_transactions.groupby("order_id").size().value_counts()
```

```python
# Question 3: Are larger baskets more common on specific days?
orders_with_products = pd.merge(products_per_order, df_train_transactions[["order_id", "order_dow"]], on="order_id")
basket_by_day = orders_with_products.groupby("order_dow")["product_count"].value_counts()
print(basket_by_day)
```

```python
# Question 4: Are reordered items more common at certain times of day?
reorder_by_hour_of_day = df_train_transactions.groupby("order_hour_of_day")["reordered"].mean()
print(reorder_by_hour_of_day)
```

```python
#Question 5: Does basket construction differ across customer order sequence?
basket_with_order_number = pd.merge(
    products_per_order,
    df_train_transactions[["order_id", "order_number"]],
    on = "order_id",
    how = "left"
)
basket_by_sequence = basket_with_order_number.groupby("order_number")["product_count"].mean()
print(basket_by_sequence)
```

### P4: Questions unlocked after final merge(train_transactions)

```python
# Creating our final merge:
df_train_analytic = pd.merge(
    df_train_transactions,
    df_products_enriched,
    on = "product_id",
    how = "left"
)
df_train_analytic.head()
```

```python
df_train_analytic.columns
```

```python
#Question 1: Which products are purchased most often?
most_purchased_products = df_train_analytic["product_name"].value_counts()
print(most_purchased_products)
```

```python
# Question 2: Which departments dominate train orders?
department_counts = df_train_analytic["department"].value_counts()
print(department_counts)
```

```python
# Question 3: Which aisles dominate train orders?
aisles_count = df_train_analytic["aisle"].value_counts()
print(aisles_count)
```

```python
# Question 4: Which departments have the highest reorder rates?
departments_by_reorder_rate = df_train_analytic.groupby("department")["reordered"].mean().sort_values(ascending = False)
print(departments_by_reorder_rate)
```

```python
# Question 5: Which aisles appear most often in large baskets?
basket_sizes = df_train_transactions.groupby("order_id")["product_id"].count().reset_index()
basket_sizes.columns = ["order_id", "basket_size"]
large_baskets = basket_sizes[basket_sizes["basket_size"] > 10]
print(large_baskets)
```

```python
large_basket_products = pd.merge(
    large_baskets,
    df_train_transactions,
    on="order_id",
    how="inner"
)
print(large_basket_products)
```

```python
large_basket_products = pd.merge(
    large_basket_products,
    df_products[["product_id", "aisle_id"]],
    on = "product_id",
    how = "left"
)
```

```python
large_basket_products = pd.merge(
    large_basket_products,
    df_aisles,
    on = "aisle_id",
    how = "left"
)
```

```python
aisles_in_large_baskets = large_basket_products["aisle"].value_counts()
print(aisles_in_large_baskets)
```

```python
#Question 6: At what times are certain departments purchased more frequently?
department_by_hour = df_train_analytic.groupby(["order_hour_of_day", "department"])["product_id"].count()
department_by_hour = department_by_hour.reset_index().sort_values(["order_hour_of_day", "product_id"], ascending=[True,False])
print(department_by_hour)
```
