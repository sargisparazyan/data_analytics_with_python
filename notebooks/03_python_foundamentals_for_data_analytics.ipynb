## Part 3

“Standard” if 100 < revenue ≤ 250
“Low” otherwise
1) Create a list of segment labels.
2) Count how many customers fall into each segment.
3) Add Markdown explanation: Why does the order of elif statements matter?

```python
segments = []

for s in sales:
    if s > 250:
        segments.append("Premium")
    elif 100 < s <= 250:
        segments.append("Standard")
    else:
        segments.append("Low")

segments
```

```python
premium_count = 0
standard_count = 0
low_count = 0
  
for segment in segments:
  if segment == "Premium":
   premium_count += 1
  if segment == "Standard":
   standard_count += 1
  if segment == "Low":
   low_count += 1
premium_count, standard_count, low_count
```

The order matters because Python checks conditions from top to bottom.

## Using a DataFrame is easier than manual loops because:
1) It automatically stores data in structured rows and columns.
2) No need to manually count rows or iterate through data.
3) Operations are faster and more readable.
4) Code is shorter and less error-prone.

## Difference between Mutable and Immutable Objects
1) Mutable objects (e.g., lists, dictionaries) can be changed after creation.
2) Immutable objects (e.g., integers, strings, tuples) cannot be changed; modifying them creates a new object.

## Why Loops Are Less Efficient Than Pandas Vectorization?
1) Loops process elements one by one in Python (slower).
2) Pandas vectorization uses optimized C-based operations that process entire columns at once (faster and more efficient).

## How Boolean Logic Relates to SQL WHERE?

1) Boolean expressions (>, <, ==, and, or) filter data in Python.
SQL WHERE uses the same logical conditions to filter rows.
Example:
df[df["revenue"] > 100] ≈ SELECT * FROM table WHERE revenue > 100

## Why Understanding Lists Helps Understand DataFrames?
1) A DataFrame column behaves like a list/array of values.
2) Filtering, indexing, and iteration concepts come from list logic.
3) Mastering lists makes DataFrame operations easier to understand.
